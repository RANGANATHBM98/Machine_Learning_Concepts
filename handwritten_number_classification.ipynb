# Mnist dataset whic is set of 70,000 small images of digits handwritten by high school students and employees of the US census Bureau.each image is labelled with the digit it represent

```python
import numpy as np
from sklearn.datasets import fetch_openml
mnist=fetch_openml('mnist_784',version=1,as_frame=False)
```

```python
x,y=mnist['data'],mnist['target']
```

```python
x_train,x_test,y_train,y_test=x[:60000],x[60000:],y[:60000],y[60000:]
```

```python
x_train.shape,x_test.shape,y_train.shape,y_test.shape
```

```python
y_train[:10]
```

```python
# converting the training dataset string to integer
y_train_int=y_train.astype(int)
y_test_int=y_test.astype(int)
```

```python
# 4th step
# it is good idea to scale the pixel values
```

```python
x_train_dumy=x_train/225.0
x_test_dumy=x_test/225.0
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
log_reg=LogisticRegression(max_iter=100,solver='lbfgs')
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
log_reg.fit(x_train_dumy,y_train_int)
```

```python
from sklearn.metrics import accuracy_score
```

```python
y_pred=log_reg.predict(x_test_dumy)
```

```python
accuracy=accuracy_score(y_test_int,y_pred)
```

```python
print("Accuracy:",accuracy*100)
```

```python
sample=x_test_dumy[0].reshape(1,-1)
probablities=log_reg.predict_proba(sample)
print(probablities)
```

```python
print("Actual",y_test_int[0])
print("Predicted",log_reg.predict(sample)[0])
```

```python
from sklearn.metrics import confusion_matrix
```

```python
cm=confusion_matrix(y_test_int,y_pred)
```

```python
import seaborn as sns
plt.figure(figsize=(10, 8))
sns.heatmap(cm,annot=True,fmt="d",cmap="Blues",xticklabels=range(10), yticklabels=range(10),  )
plt.xlabel("Predicted Label") 
plt.ylabel("Actual Label") 
plt.title("MNIST Confusion Matrix") 
plt.show()
```

```python
from sklearn.metrics import classification_report
```

```python
print("--- Classification Report ---")
print(
    classification_report(
        y_test_int,
        y_pred,
        target_names=[f"Digit {i}" for i in range(10)],
    )
)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
image=cv2.imread('07.png')
print(image.shape)
plt.imshow(cv2.cvtColor(image,cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

```python
#convert into grace
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)

#blur they grey image
blur=gray.copy
blur=cv2.GaussianBlur(gray,(5,5),0)
print(gray.shape)
print(blur.shape)
```

```python
#convert into grace
gray=cv2.cvtColor(image,cv2.COLOR_BGR2GRAY)

#blur they grey image
blur=gray.copy
blur=cv2.GaussianBlur(gray,(5,5),0)
#threshold
_,threshold=cv2.threshold(gray,0,255,cv2.THRESH_BINARY_INV+cv2.THRESH_OTSU)
```

```python
plt.figure(figsize=(21, 6))

plt.subplot(1, 4, 1)
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(gray, cmap="gray")
plt.title("grey image")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(blur, cmap="gray")
plt.title("blurred image")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(threshold, cmap="gray")
plt.title("threshold")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
contours,_=cv2.findContours(
    threshold,
    cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE
)
print("number of counter",len(contours))
```

```python
largest_contours=max(contours,key=cv2.contourArea)
x,y,w,h=cv2.boundingRect(largest_contours)
```

```python
print("Bounding box")
print("x",x)
print("y",y)
print("width",w)
print("height",h)
```

```python
# crop the image:
corp_digit=threshold[y:y+h,x:x+w]
#display the cropped digit
plt.figure(figsize=(4,4))
plt.imshow(corp_digit,cmap="gray")
plt.show()
```

```python
# add padding to the image
padding=20
digit_padded=cv2.copyMakeBorder(corp_digit,padding,padding,padding,padding,cv2.BORDER_CONSTANT,value=0)
plt.imshow(digit_padded,cmap='gray')
plt.show()
```

```python
# reshape the padded  image
digit_28=cv2.resize(digit_padded,(28,28),interpolation=cv2.INTER_AREA)
print("Shape:",digit_28.shape)
```

```python
plt.imshow(digit_28,cmap='gray')
plt.show()
```

```python
# final mnist stlye image 
plt.figure(figsize=(6,6))
plt.imshow(digit_28,cmap='gray',interpolation='nearest')
plt.xticks(range(0,28,2))
plt.yticks(range(0,28,2))
plt.grid()
plt.title("final mnist-style image")
plt.show()
```

```python
digit_normalised=digit_28.astype('float32')/255.0
```

```python
print("Minimum",digit_normalised.min())
print("Maximum",digit_normalised.max())
```

```python
## flattened image
digit_flattened=digit_normalised.reshape(1,784)
print('shape',digit_flattened.shape)
```
