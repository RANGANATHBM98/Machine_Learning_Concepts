# K-means Algorithm
k means clustering is an unsupervised machine learning algorithm which group the unlabelled dataset into different clusters.Here k defines the number of predefined clusters that need to be created in the process as if k=2 then it means there 2 cluster and if it 3 it means there is 3 cluster

## the k means clustering algorithm mainly perform 2 tasks:-
1.determine best value for center points or centorids by an iterative process                       
2.aaign each data points to its closest k center .those data point which are near to particular k-center,that point merge to center


# elbow method 

The elbow methods is one of the most popular ways to find the optimal number of clusters, this method uses the concept of WCSS value. Where WCSS is stands for "Within cluster of squares" Which defines the total variation within a cluster

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
from pylab import rcParams
rcParams['figure.figsize']=10,8
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
#reading of the csv file 
c1=pd.read_csv('clustering_data_1.csv')
c1.head()
```

```python
c1.shape
```

```python
plt.scatter(c1['X'], c1['Y'], marker='.')
plt.xlabel("$X_Axis$", fontsize=16)
plt.ylabel("$Y_Axis$", rotation=0, fontsize=16)
plt.title("data in the scatter")
plt.show()
```

```python
from sklearn.cluster import KMeans
```

```python
kmeans = KMeans(n_clusters=2, random_state=42)
```

```python
kmeans.fit(c1)
```

```python
cluster_assignment=kmeans.predict(c1)
```

```python
plt.scatter(c1['X'], c1['Y'], marker='.',c=cluster_assignment)
plt.xlabel("$X_Axis$", fontsize=16)
plt.ylabel("$Y_Axis$", rotation=0, fontsize=16)
plt.title("data in the scatter")
plt.show()
```

```python
kmeans_3=KMeans(n_clusters=3,random_state=42)
```

```python
kmeans_3.fit(c1)
```

```python
cluster_assignment=kmeans_3.predict(c1)
```

```python
plt.scatter(c1['X'], c1['Y'], marker='.',c=cluster_assignment)
plt.xlabel("$X_Axis$", fontsize=16)
plt.ylabel("$Y_Axis$", rotation=0, fontsize=16)
plt.title("data in the scatter")
plt.show()
```

```python
c1.head()
```

```python
print(kmeans_3.labels_)
```

```python
y_predict=kmeans_3.predict(c1[['X','Y']])
y_predict
```

```python
c1['cluster_prediction']=y_predict
c1.head()
```

```python
c1['cluster_prediction'].value_counts()
```

```python
c1['cluster_prediction'].value_counts().plot(kind='pie',autopct='%.2f%%',cmap='Set3')
```

```python
cluster_assignment=cluster_assignment.astype(str)
cluster_assignment[cluster_assignment=='0']='lightblue'
cluster_assignment[cluster_assignment=='1']='yellow'
cluster_assignment[cluster_assignment=='2']='skyblue'
plt.scatter(c1['X'],c1['Y'],marker='.',c=cluster_assignment)
plt.show()
```

```python
#centroid value
kmeans_3.cluster_centers_
```

```python
df1=c1[c1.cluster_prediction==0]
df2=c1[c1.cluster_prediction==1]
df3=c1[c1.cluster_prediction==2]
```

```python
plt.scatter(df1.X,df1.Y,color='Green',marker='.',label='cluster 0')
plt.scatter(df2.X,df2.Y,color='Red',marker='^',label='cluster 1')
plt.scatter(df3.X,df3.Y,color='Blue',marker='1',label='cluster 2')
plt.scatter(kmeans_3.cluster_centers_[:,0],kmeans_3.cluster_centers_[:,1],color='black',marker='*',label='centroid')
plt.xlabel("X")
plt.ylabel("Y",rotation=False)
plt.legend()
plt.show()
```

```python
#plotting of elbow
sse=[]
k_ran=range(1,10)
for k in k_ran:
    km=KMeans(n_clusters=k)
    km.fit(c1[['X','Y']])
    sse.append(km.inertia_)
plt.xlabel('K')
plt.ylabel('Sum of squared number')
plt.plot(k_ran,sse)
```

```python
from sklearn.datasets import make_moons
from sklearn.cluster import SpectralClustering,KMeans
from sklearn.metrics import adjusted_rand_score
```

```python
X,y=make_moons(n_samples=300,noise=0.05,random_state=42)
```

```python
sc=SpectralClustering(n_clusters=2,affinity='nearest_neighbors',random_state=42,n_init=10)
```

```python
sc.fit(X)
```

```python
labels_sc=sc.fit_predict(X)
```

```python
km=KMeans(n_clusters=2,random_state=42,n_init=10)
```

```python
labels_km=km.fit_predict(X)
```

```python
# compare and check which fails
print("Spectural ARI:",adjusted_rand_score(y,labels_sc))
print("KMEANS ARI:",adjusted_rand_score(y,labels_km))
```

```python
# Plotting
fig,axes=plt.subplots(ncols=1,nrows=3,figsize=(15,4))
axes[0].scatter(X[:,0],X[:,1],c=y,cmap='viridis')
axes[0].set_title("ground Truth")
axes[1].scatter(X[:,0],X[:,1],c=labels_sc,cmap='viridis')
axes[1].set_title("SpectralClustering")
axes[2].scatter(X[:,0],X[:,1],c=labels_km,cmap='viridis')
axes[2].set_title("k_means(Fails")
plt.show()
```
